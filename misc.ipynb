```python
import pmdarima as pm

model = pm.auto_arima(
    y_train,
    start_p=1, max_p=2,      # we know p is ~1-2, don't waste time checking 0
    d=0,
    start_q=0, max_q=1,
    start_P=0, max_P=1,
    D=1,                      # confirmed by our stationarity check
    start_Q=1, max_Q=2,       # we found strong evidence for seasonal MA
    m=24,
    trace=True,
    stepwise=True
)
print(model.summary())
```

```python
candidates = [
    ((1,0,1), (1,1,0,24)),
    ((2,0,0), (1,1,0,24)),
    ((2,0,1), (1,1,0,24)),
    ((1,0,0), (1,1,0,24)),
]

results = {}
for order, seasonal_order in candidates:
    m = pm.ARIMA(order=order, seasonal_order=seasonal_order, suppress_warnings=True)
    m.fit(y_train)
    results[(order, seasonal_order)] = m.aic()
    print(order, seasonal_order, m.aic())
    del m  # free memory before next fit
```
